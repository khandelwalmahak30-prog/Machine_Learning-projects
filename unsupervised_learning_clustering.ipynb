```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
data= {
    'customer:':['riya','aman','sakshay','swati','raman','sneha'],
    'age':[20,30,40,26,32,18],
    'spending':[100,200,300,110,120,290]
}
df=pd.DataFrame(data)
X=df[['age','spending']]
model=KMeans(n_clusters=2,random_state=42,n_init=10)
## n init: no of times clustering and ncluster : 2 bnane he and random:har bar same result aaye
df['group']=model.fit_predict(X) ##grp nam ka column add kr dega table me

plt.figure(figsize=(6,5))
for group in df['group'].unique():##unique: hr colour me plot kr ske [0,1]me grp nikalo
  group_data=df[df['group']==group] ##sirf current ka grp ka data nikalo
  plt.scatter(group_data['age'],group_data['spending'],label=f'group{group}')

plt.xlabel('age')
plt.ylabel('sepnding score')
plt.title('customer segments (K-Means)')
plt.legend()
plt.grid(True)
plt.show()
print(df)
```

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
data={
    'customer':['rahul','priya','aman','tina','rohan','pooja'],
    'orders':[2,3,15,18,1,20],
    'amount':[200,300,1500,1800,100,2000]
}
df=pd.DataFrame(data)
X=df[['orders','amount']]
model=KMeans(n_clusters=3,random_state=42,n_init=10)
df['group']=model.fit_predict(X)

plt.figure(figsize=(6,5))
for group in df['group'].unique():
  group_data=df[df['group']==group]
  plt.scatter(group_data['orders'],group_data['amount'],label=f'group{group}')
plt.xlabel('orders')
plt.ylabel('amount')
plt.title('food orders : KMEANS')
plt.legend()
plt.grid(True)
plt.show()
print(df)
```
